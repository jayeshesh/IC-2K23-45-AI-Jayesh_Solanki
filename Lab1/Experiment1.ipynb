## Step 2 - Detect the runtime (Jupyter or Google Colab)
`get_ipython()` exists only inside an IPython/Jupyter kernel; the module `google.colab` exists only on Google Colab.

````python
def detect_runtime() -> str:
    """Return a short description of the execution runtime."""
    try:
        shell = get_ipython().__class__.__name__
    except NameError:
        return 'Plain Python interpreter'
    if 'google.colab' in sys.modules:
        return 'Google Colab (' + shell + ')'
    return 'Local Jupyter kernel (' + shell + ')'


print('Runtime:', detect_runtime())
````

## Step 4 - Python constructs used to represent AI problems
| Construct | Use in AI problem solving |
|---|---|
| `tuple` | immutable, hashable **state**, e.g. `(x, y)` for the water jugs |
| `set` / `dict` | record of already generated states, parent links |
| `collections.deque` | FIFO list of states waiting to be processed |
| functions / lambdas | **production rules** (condition and action) |

````python
from collections import deque

CAP_A, CAP_B = 4, 3
state = (0, 0)                      # tuple -> hashable state

# A production rule = (name, condition, action)
fill_a = ('Fill Jug A',
          lambda x, y: x < CAP_A,
          lambda x, y: (CAP_A, y))

name, condition, action = fill_a
if condition(*state):
    new_state = action(*state)
    print(f'{name}: {state} -> {new_state}')

seen = {state, new_state}           # set of generated states
pending = deque([new_state])        # FIFO queue
print('Generated states :', sorted(seen))
print('Queue contents   :', list(pending))
print('Tuple hashable   :', hash(state) is not None)
````
